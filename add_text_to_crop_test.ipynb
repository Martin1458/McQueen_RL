```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

load image

```python
# load image
image_path = "./portrait_better.png"
frame = cv2.imread(image_path)

# get details
height, width = frame.shape[:2]
print(f"Image dimensions: {width} x {height}")

# show image
plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

adding text

```python
# add text to the top
text_line1 = "Full game to let you"
text_line2 = "rest from doomscrolling"
cv2.putText(frame, text_line1, (120, 120), cv2.FONT_HERSHEY_SIMPLEX, 3, (255, 255, 255), 2)
cv2.putText(frame, text_line2, (30, 220), cv2.FONT_HERSHEY_SIMPLEX, 3, (255, 255, 255), 2)

# show image with text
plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
